# 3D C-Space NDF Visualization
This notebook provides interactive visualization and analysis of the learned 3D Configuration Space (C-Space) for the Jaco arm.

```python
import os
import torch
import numpy as np
import matplotlib.pyplot as plt
from skimage import measure
from utils import load_model

# Enable interactive plots if in a compatible environment
%matplotlib inline
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")
```

```python
model_path = "output/ndf/model_final.pt"
if not os.path.exists(model_path):
    print(f"Error: {model_path} not found. Please run training first.")
else:
    print(f"Loading model from {model_path}...")
    model = load_model(model_path, device)
```

```python
res = 64
ticks = np.linspace(-np.pi, np.pi, res)
q1, q2, q3 = np.meshgrid(ticks, ticks, ticks, indexing='ij')
grid_points = np.stack([q1.ravel(), q2.ravel(), q3.ravel()], axis=1)

print(f"Predicting on {res}^3 grid...")
with torch.no_grad():
    batch_size = 100000
    preds = []
    for i in range(0, len(grid_points), batch_size):
        batch = torch.Tensor(grid_points[i:i+batch_size]).to(device)
        p = model(batch).cpu().numpy()
        preds.append(p)
    Z = np.concatenate(preds).reshape(res, res, res)
print("Done.")
```

### 3D Scatter Plot (C-Obstacle)
Points where predicted distance is less than 0 represent collisions.

```python
fig = plt.figure(figsize=(10, 8))
ax = plt.axes(projection='3d')

mask = Z < 0
indices = np.where(mask)

max_scatter = 15000
if len(indices[0]) > max_scatter:
    idx = np.random.choice(len(indices[0]), max_scatter, replace=False)
    xs, ys, zs = ticks[indices[0][idx]], ticks[indices[1][idx]], ticks[indices[2][idx]]
else:
    xs, ys, zs = ticks[indices[0]], ticks[indices[1]], ticks[indices[2]]

ax.scatter(xs, ys, zs, c='red', s=1, alpha=0.2)
ax.set_title("Learned 3D C-Space (Collision Points)")
ax.set_xlabel("q1"); ax.set_ylabel("q2"); ax.set_zlabel("q3")
ax.set_xlim(-np.pi, np.pi); ax.set_ylim(-np.pi, np.pi); ax.set_zlim(-np.pi, np.pi)
plt.show()
```

### 2D Slices ($q_3$ slices)

```python
n_slices = 9
slice_indices = np.linspace(0, res-1, n_slices, dtype=int)
fig, axes = plt.subplots(3, 3, figsize=(12, 12))
axes = axes.flatten()

for i, idx in enumerate(slice_indices):
    ax = axes[i]
    slice_data = Z[:, :, idx]
    ax.imshow(slice_data.T, origin='lower', extent=[-np.pi, np.pi, -np.pi, np.pi], cmap='RdBu', vmin=-0.5, vmax=0.5)
    ax.contour(ticks, ticks, slice_data.T, levels=[0], colors='black', linewidths=1)
    ax.set_title(f"q3 = {ticks[idx]:.2f} rad")
    ax.set_xlabel("q1"); ax.set_ylabel("q2")

plt.tight_layout()
plt.show()
```

### Mesh Extraction (OBJ Export)
Generate a smooth surface using Marching Cubes.

```python
try:
    verts, faces, normals, values = measure.marching_cubes(Z, level=0)
    verts = (verts / (res - 1)) * (2 * np.pi) - np.pi
    
    obj_path = "output/ndf/final_3d_viz.obj"
    with open(obj_path, 'w') as f:
        for v in verts: f.write(f"v {v[0]} {v[1]} {v[2]}\n")
        for face in faces: f.write(f"f {face[0]+1} {face[1]+1} {face[2]+1}\n")
    print(f"Successfully saved mesh to {obj_path}")
except Exception as e:
    print(f"Mesh extraction failed: {e}")
```
